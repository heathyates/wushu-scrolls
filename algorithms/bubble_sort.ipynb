```python
def bubble_sort(self, array): 
        """
        Sorts an integer list using Bubble Sort.
        """
        try:
            n = len(array)
            for i in range(n): 
                swap = False 
                for j in range(n - i - 1):  # Compare adjacent elements
                    if array[j] > array[j + 1]:  # Swap if left is greater
                        array[j], array[j + 1] = array[j + 1], array[j]
                        swap = True
                if not swap:  # If no swaps occur, array is already sorted
                    break
        except Exception as e:
            return "derp"
        
        return array
```
